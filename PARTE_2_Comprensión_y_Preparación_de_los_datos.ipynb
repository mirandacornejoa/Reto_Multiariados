## Preparación de los datos

Columnas objetivo: PM2.5, PM10, NO₂, NOx, SO₂ y CO

```python
# Eliminar filas donde 'date' es NaT
df2023.dropna(subset=['date'], inplace=True)
df2024.dropna(subset=['date'], inplace=True)
df2025.dropna(subset=['date'], inplace=True)

print("Filas con 'date' faltante eliminadas:")
print(df2023.isnull().sum())
print(df2024.isnull().sum())
print(df2025.isnull().sum())
```

```python
# Corrige valores érroneos
df2025.replace('Nulo', np.nan, inplace=True)
df2024.replace('Nulo', np.nan, inplace=True)
df2023.replace('Nulo', np.nan, inplace=True)
```

```python
# Maneja valores faltantes con imputación media movil
# Para df2025
num_cols = df2025.select_dtypes(include=["float64", "int64"]).columns

for col in num_cols:
    rolling_mean = df2025[col].rolling(window=5, min_periods=1).mean()
    df2025[col] = df2025[col].fillna(rolling_mean)

# Para df2024
num_cols = df2024.select_dtypes(include=["float64", "int64"]).columns

for col in num_cols:
    rolling_mean = df2024[col].rolling(window=5, min_periods=1).mean()
    df2024[col] = df2024[col].fillna(rolling_mean)

# Para df2023
num_cols = df2023.select_dtypes(include=["float64", "int64"]).columns

for col in num_cols:
    rolling_mean = df2023[col].rolling(window=5, min_periods=1).mean()
    df2023[col] = df2023[col].fillna(rolling_mean)
```

```python
# Interpolación
df2025[col] = df2025[col].fillna(rolling_mean).interpolate()
df2024[col] = df2024[col].fillna(rolling_mean).interpolate()
df2023[col] = df2023[col].fillna(rolling_mean).interpolate()
```

```python
# Manejo de outliers
# --- RANGOS DE OPERACIÓN POR AÑO (según SIMA) ---
rangos_operacion = {
    2023: {
        'PM10': (0, 900), 'PM2.5': (0, 800), 'O3': (0, 175), 'NO': (0, 500),
        'NO2': (0, 175), 'NOX': (0, 500), 'SO2': (0, 250), 'CO': (0, 14),
        'RH': (0, 100), 'WSR': (0, 40), 'TOUT': (0, 45), 'SR': (0, 1),
        'PRS': (690, 740), 'WDR': (0, 360), 'RAINF': (0, 70)
    },
    2024: {
        'PM10': (0, 999), 'PM2.5': (0, 999), 'O3': (0, 180), 'NO': (0, 400),
        'NO2': (0, 130), 'NOX': (0, 500), 'SO2': (0, 150), 'CO': (0, 18),
        'RH': (0, 100), 'WSR': (0, 38), 'TOUT': (-4, 45.5), 'SR': (0, 1.26),
        'PRS': (687.5, 740), 'WDR': (0, 360), 'RAINF': (0, 50)
    },
    2025: {
        'PM10': (0, 820), 'PM2.5': (0, 350), 'O3': (0, 185), 'NO': (0, 350),
        'NO2': (0, 175), 'NOX': (0, 400), 'SO2': (0, 405), 'CO': (0, 10),
        'RH': (0, 100), 'WSR': (0, 40), 'TOUT': (-4.5, 45), 'SR': (0, 1.2),
        'PRS': (688, 740), 'WDR': (0, 360), 'RAINF': (0, 25)
    }
}

# --- FUNCIÓN GENERAL DE LIMPIEZA ---
def limpiar_outliers_operacion(df, year):
    """
    Elimina (pone en NaN) todos los valores fuera del rango de operación del SIMA para el año dado.
    """
    df_clean = df.copy()
    rangos = rangos_operacion.get(year, None)
    if rangos is None:
        raise ValueError(f"No hay rangos definidos para el año {year}")

    print(f"\n=== Limpieza de outliers según rangos de operación {year} ===")
    for col, (min_val, max_val) in rangos.items():
        if col in df_clean.columns:
            mask_out = (df_clean[col] < min_val) | (df_clean[col] > max_val)
            n_outliers = mask_out.sum()
            if n_outliers > 0:
                print(f"  {col}: {n_outliers} fuera de rango ({min_val}, {max_val}) → reemplazados por NaN")
                df_clean.loc[mask_out, col] = np.nan
    print("============================================================\n")
    return df_clean

# --- APLICAR A CADA AÑO ---
df2023 = limpiar_outliers_operacion(df2023, 2023)
df2024 = limpiar_outliers_operacion(df2024, 2024)
df2025 = limpiar_outliers_operacion(df2025, 2025)
```

## Transformación de datos

Discretizacion de datos no es necesario realizarlo

```python
def add_time_features(df, date_col='date'):
    dt = df[date_col].dt

    df['weekday'] = dt.weekday
    df['month'] = dt.month

    month = df['month']
    df['season'] = np.select(
        [
            month.isin([12,1,2]),
            month.isin([3,4,5]),
            month.isin([6,7,8]),
            month.isin([9,10,11])
        ],
        [0,1,2,3]  # 0=inv, 1=prim, 2=ver, 3=oto
    )
    return df

df_final = add_time_features(df_final)
df_final.drop(columns="month")
```

```python
def add_wind_components(df, speed_col='WSR', dir_col='WDR'):
    df = df.copy()
    df['WSR_ms'] = df['WSR'] * (1000/3600)  # km/h -> m/s

    # Convertir a radianes; para meteo, u = -V * sin(theta), v = -V * cos(theta)
    theta = np.deg2rad(df[dir_col].astype(float))
    V = df['WSR_ms'].astype(float)
    df['wind_u'] = -V * np.sin(theta)
    df['wind_v'] = -V * np.cos(theta)
    return df
df_final = add_wind_components(df_final)
df_final.drop(columns="WSR_ms")
```

```python
def add_chemistry_features(df, no_col='NO', no2_col='NO2', nox_col='NOX', o3_col='O3', sr_col='SR'):
    df = df.copy()

    # Define a small epsilon value to avoid division by zero
    _EPS = 1e-9

    # Razones fotoquímicas

    # Índices simples útiles en práctica:
    # - Photo index (ozono por radiación / precursores)
    if sr_col in df.columns:
        df['photo_index'] = df[o3_col] * (df[sr_col] + _EPS) / (df[nox_col] + _EPS)

    return df
df_final = add_chemistry_features(df_final)
df_final.head()
```

```python
# Escalar y normalizar los datos
import pandas as pd
from sklearn.preprocessing import RobustScaler

# Variables numéricas (según tu lista)
NUMERIC_COLS = [
    'CO','NO','NO2','NOX','O3','PM10','PM2.5',
    'PRS','RAINF','RH','SO2','SR','TOUT','WSR','WDR'
]

PASSTHROUGH_COLS = ['date', 'Estacion']  # estas no se escalan

# Función para aplicar RobustScaler
def robust_scale_df(df, numeric_cols=NUMERIC_COLS, passthrough_cols=PASSTHROUGH_COLS):
    scaler = RobustScaler()
    df_scaled = df.copy()

    # Ajustar y transformar solo variables numéricas
    df_scaled[numeric_cols] = scaler.fit_transform(df[numeric_cols])

    return df_scaled, scaler  # regreso también el scaler por si quieres guardarlo

# Si en lugar de hacerlo independiente quieres un escalado común:
# Combina primero, ajusta una sola vez y luego divide
scaler = RobustScaler()
df_final[NUMERIC_COLS] = scaler.fit_transform(df_final[NUMERIC_COLS])
```
